```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Name": ["Puja", "Riya", "Anchal", "Nidhi", "Bhoomika", "Khushi"],
    "Department": ["BCA", "BCA", "BBA", "BCA", "BBA", "BBA"],
    "Attendance": [85, 92, 70, 88, 65, 95],
    "Study-Hours": [4, 6, 2, 5, 2, 7],
    "Python": [82, 95, 58, 80, 45, 96],
    "DBMS": [75, 90, 62, 78, 48, 92],
    "AI": [80, 94, 60, 85, 52, 95]
}

df = pd.DataFrame(data)

df["Average"] = df[["Python", "DBMS", "AI"]].mean(axis=1)

print("STUDENT PERFORMANCE")
print(df[["Name", "Department", "Attendance",
          "Study-Hours", "Average"]].round(2))

print("\nHIGH PERFORMING STUDENTS")
print(df[df["Average"] >= 80][["Name", "Average"]].round(2))

print("\nLOW PERFORMING STUDENTS")
print(df[df["Average"] < 60][["Name", "Average"]].round(2))

print("\nSUBJECT-WISE AVERAGE")
print(df[["Python", "DBMS", "AI"]].mean().round(2))

print("\nDEPARTMENT-WISE PERFORMANCE")
print(df.groupby("Department")["Average"].mean().round(2))

plt.scatter(df["Study-Hours"], df["Average"])
plt.xlabel("Study Hours")
plt.ylabel("Average Marks")
plt.title("Study Hours vs Average Marks")
plt.show()

plt.scatter(df["Attendance"], df["Average"])
plt.xlabel("Attendance")
plt.ylabel("Average Marks")
plt.title("Attendance vs Performance")
plt.show()
```
